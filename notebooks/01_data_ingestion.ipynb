```python
# Cell 1: Environment Setup & Library Import
import sys
from pathlib import Path

# Add project root directory to path
sys.path.append(str(Path.cwd().parent))

# Import functions directly from the src package
from src import load_data, clean_and_impute, create_binary_target
```

```python
# Cell 2: Execute Step 1 Data Pipeline
RAW_PATH = "../data/fault_data.csv"
PROCESSED_PATH = "../data/proc_fault_data.csv"

# 1. Load Dataset
df_raw = load_data(RAW_PATH)
print(f"Loaded dataset with shape: {df_raw.shape}")

# 2. Impute Missing Values
df_clean = clean_and_impute(df_raw)

# 3. Create Binary Target (is_critical)
df_processed = create_binary_target(df_clean, target_col='Component Health')

# 4. Save Processed CSV
df_processed.to_csv(PROCESSED_PATH, index=False)
print(f"Step 1 Complete! Saved processed dataset to: {PROCESSED_PATH}")
```
