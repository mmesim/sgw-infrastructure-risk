```python
# Cell 1: Environment Setup & Library Import
import sys
from pathlib import Path

# Add project root directory to path
sys.path.append(str(Path.cwd().parent))

# Import functions from the src package
from src import load_data, encode_categorical_features, engineer_risk_features, prepare_feature_matrix
```

```python
# Cell 2: Execute Step 2 Feature Pipeline
PROCESSED_PATH = "../data/proc_fault_data.csv"

# 1. Load processed data from Step 1
df = load_data(PROCESSED_PATH)
print(f"Loaded processed data with shape: {df.shape}")

# 2. Encode Categorical Variables
df_encoded = encode_categorical_features(df, cat_cols = ['Weather Condition', 'Maintenance Status'])

print(f"Encoded data with shape: {df_encoded.shape}")
#show the new columns 
df_encoded.columns
```

```python
# 3. Engineer Domain Features
df_featured = engineer_risk_features(df_encoded)

# 4. Prepare X (Features) and y (Target) for XGBoost
# Pass an extra list of columns you want to drop during matrix creation
X, y = prepare_feature_matrix(df_featured, target_col='is_critical', exclude_cols=['Fault ID', 'Component Health', 'Asset ID'])

print(f"Feature Matrix (X) Shape: {X.shape}")
print(f"Target Vector (y) Shape: {y.shape}")
print("Feature columns ready for modeling:", (X.columns))
print("Target column ready for modeling:", y.name)
```
